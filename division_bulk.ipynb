```python
import pandas as pd
import math

def split_csv_for_canva(master_filename):
    print(f"Loading final master data from {master_filename}...")
    
    try:
        df = pd.read_csv(master_filename)
    except FileNotFoundError:
        print(f"Error: Could not find {master_filename}")
        return

    df = df.head(45)
    
    rows_per_chunk = math.ceil(len(df) / 3)
    
    for i in range(3):
        start_row = i * rows_per_chunk
        end_row = (i + 1) * rows_per_chunk
        
        chunk = df.iloc[start_row:end_row]
        
        # Save to CSV
        output_name = f"Canva_Layout_{i+1}_Data.csv"
        chunk.to_csv(output_name, index=False, encoding='utf-8')
        print(f"Saved {len(chunk)} articles to {output_name}")

    print("\nSUCCESS! You now have 3 CSVs ready for Canva Bulk Create.")

if __name__ == "__main__":
    MASTER_CSV = "Magazine_50_Pages_Final_Images.csv" 
    split_csv_for_canva(MASTER_CSV)
```
