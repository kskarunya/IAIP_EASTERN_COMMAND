```python
import pandas as pd
from urllib.parse import urlparse

def generate_references_appendix(input_filename, output_text_file):
    print(f"Loading data from {input_filename} to generate Appendix...")
    
    try:
        df = pd.read_csv(input_filename)
    except FileNotFoundError:
        print(f"Error: Could not find {input_filename}")
        return

    with open(output_text_file, 'w', encoding='utf-8') as f:
        f.write("APPENDIX B: SOURCE DIRECTORY & REFERENCES\n")
        f.write("="*50 + "\n\n")
        
        if 'Category' in df.columns:
            df = df.sort_values(by='Category')
            
        current_category = ""
        
        for index, row in df.iterrows():
            title = row.get('Title', 'Unknown Title')
            url = row.get('URL', 'No URL provided')
            category = row.get('Category', 'General')
            
            if category != current_category:
                f.write(f"\n--- {category.upper()} ---\n")
                current_category = category
                
            try:
                domain = urlparse(url).netloc.replace('www.', '')
            except:
                domain = "News Source"
                
            f.write(f"{index + 1}. {domain.upper()} - \"{title}\"\n")
            f.write(f"   Link: {url}\n\n")

    print(f"SUCCESS! Appendix B has been generated and saved to: {output_text_file}")

if __name__ == "__main__":
    INPUT_FILE = "Magazine_50_Pages_Final_Images.csv" 
    OUTPUT_FILE = "Appendix_B_References.txt"
    
    generate_references_appendix(INPUT_FILE, OUTPUT_FILE)
```
